Welcome to the **Agentic Analytics** workshop. You'll build a text-to-SQL BI agent that answers
natural-language questions about a fictional university's student data — by writing SQL, running it

## What we're building (the whole picture)

![Architecture](assets/architecture.png)

By the end of the workshop the system looks like the diagram above:

| Component | Role |
|-----------|------|
| **AgentCore runtime w/ Claude agent** | the brain — turns a question into SQL, runs it, analyzes results (Modules 1–3) |
| **Amazon Bedrock** | serves the Claude model the agent runs on |
| **Amazon Athena** | the SQL query engine (serverless — queries data directly in S3) |
| **Amazon S3** | stores the raw data **and** the query results |

**Module 0 builds only the data layer** — the S3 bucket, the data, and the Athena
database/tables. The agent and its runtime come in Modules 1–3. We set up the data first so the
agent has something real to query.

## Create the infrastructure

A single idempotent script (`scripts/setup_infrastructure.py`) does all of this — safe to re-run:

1. an **S3 bucket** `student-analytics-agent-<your-account-id>` (data + Athena results),
2. uploads the two **demo CSVs** (≈60,000 rows of student enrollment + financial data),
3. an **Athena database** `student_analytics` with two external tables over that data,
4. a **smoke query** to prove it all works.

Run the two cells below: the first detects your AWS region, and the second creates the infrastructure.

````python
import os
from dotenv import load_dotenv
load_dotenv()

REGION = os.getenv("AWS_REGION", "us-east-1")
os.environ["AWS_REGION"] = REGION
print(f"✅ Region: {REGION}")
````

````python
# One command stands up everything (idempotent — safe to re-run).
!uv run python scripts/setup_infrastructure.py --region {REGION}
````
